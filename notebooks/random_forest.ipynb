## Setup

```python
import pandas as pd
import numpy as np
from sentence_transformers import SentenceTransformer
from functions import *
from sklearn.ensemble import RandomForestRegressor
```

```python
# import data
df_train = pd.read_csv('../data/raw/neurips-open-polymer-prediction-2025/train.csv')
df_test = pd.read_csv('../data/raw/neurips-open-polymer-prediction-2025/test.csv')
print(f'Train shape: {df_train.shape}')
print(f'Test shape: {df_test.shape}')
```

```python
# Now let us create embeddings for each SMILES string.
# st_model = SentenceTransformer("Derify/ChemMRL-alpha")
# using the model below as on Kaggle I was able to use that only. Was facing some difficulty in uploading the Derify/ChemMRL-alpha model on kaggle
st_model = SentenceTransformer('/Users/shashankkatiyar/Documents/learning_ml/open_polymer_prediction/models/ChemBERTa-77M-MLM')
df_train['embedding'] = [st_model.encode(x) for x in df_train['SMILES']]
```

```python
# get all the subsets of the df, where each subset has all the information for one property
data_tg = df_train.loc[~df_train['Tg'].isna(), ['id', 'SMILES', 'Tg', 'embedding']]
data_ffv = df_train.loc[~df_train['FFV'].isna(), ['id', 'SMILES', 'FFV', 'embedding']]
data_tc = df_train.loc[~df_train['Tc'].isna(), ['id', 'SMILES', 'Tc', 'embedding']]
data_density = df_train.loc[~df_train['Density'].isna(), ['id', 'SMILES', 'Density', 'embedding']]
data_rg = df_train.loc[~df_train['Rg'].isna(), ['id', 'SMILES', 'Rg', 'embedding']]
```

## Hyperparameter Optimization (no need to run again)

### Tg

```python
X = np.vstack(data_tg['embedding'])
y = data_tg['Tg']
model_tg = OptimizeHyperparamsRF(X, y)
```

### FFV

```python
X = np.vstack(data_ffv['embedding'])
y = data_ffv['FFV']
model_ffv = OptimizeHyperparamsRF(X, y, n_iter = 5)
```

### Tc

```python
X = np.vstack(data_tc['embedding'])
y = data_tc['Tc']
model_tc = OptimizeHyperparamsRF(X, y)
```

### Density

```python
X = np.vstack(data_density['embedding'])
y = data_density['Density']
model_density = OptimizeHyperparamsRF(X, y)
```

### Rg

```python
X = np.vstack(data_rg['embedding'])
y = data_rg['Rg']
model_rg = OptimizeHyperparamsRF(X, y)
```

## Training final model

```python
# Final parameters of each model:
models_params_dict = {
    'Tg': {'max_depth':15, 'min_samples_leaf':4, 'min_samples_split':6, 'n_estimators':149},
    'FFV': {'max_depth':15, 'min_samples_leaf':4, 'min_samples_split':6, 'n_estimators':149},
    'Tc': {'max_depth':15, 'min_samples_leaf':4, 'min_samples_split':6, 'n_estimators':70},
    'Density': {'max_depth':12, 'min_samples_leaf':4, 'min_samples_split':4, 'n_estimators':199},
    'Rg': {'max_depth':12, 'min_samples_leaf':4, 'min_samples_split':4,'n_estimators':199}
}
```

```python
model_fitted_tg = TrainRFModel(models_params_dict, df_train, 'Tg')
model_fitted_ffv = TrainRFModel(models_params_dict, df_train, 'FFV')
model_fitted_tc = TrainRFModel(models_params_dict, df_train, 'Tc')
model_fitted_density = TrainRFModel(models_params_dict, df_train, 'Density')
model_fitted_rg = TrainRFModel(models_params_dict, df_train, 'Rg')
```

```python
# compile the models in a single dict
models_dict = {
    'Tg': model_fitted_tg,
    'FFV': model_fitted_ffv,
    'Tc': model_fitted_tc,
    'Density': model_fitted_density,
    'Rg': model_fitted_rg
}
```
